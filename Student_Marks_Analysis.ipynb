# Student Marks Analysis
## Task 1

This project analyzes student marks using Python.

### Objectives
- Calculate total marks
- Calculate average marks
- Find highest marks
- Find lowest marks

### Tools
- Python
- Jupyter Notebook

```python
# Store student marks in a Python list

marks = [78, 85, 92, 67, 88, 74, 95, 81, 69, 90]

print("Student Marks:", marks)
```

## 1. Total Marks

The `sum()` function is used to calculate the total of all student marks.

```python
# Calculate total marks
total_marks = sum(marks)

# Calculate total marks
total_marks = sum(marks)

print("Total Marks:", total_marks)
```

## 2. Average Marks

The average is calculated by dividing the total marks by the number of students.

```python
average = total_marks / len(marks)

print("Average Marks:", round(average, 2))
```

## 3. Highest Marks

The `max()` function is used to find the highest mark in the list.

```python
highest = max(marks)

print("Highest Marks:", highest)
```

## 4. Lowest Marks

The `min()` function is used to find the lowest mark in the list.

```python
lowest = min(marks)

print("Lowest Marks:", lowest)
```

## Final Results

```python
print("===== Student Marks Analysis =====")
print("Number of Students:", len(marks))
print("Total Marks:", total_marks)
print("Average Marks:", round(average, 2))
print("Highest Marks:", highest)
print("Lowest Marks:", lowest)
```
